```python
df['TotalRevenue'] = df['TotalRevenue'].fillna(0)
```

```python
df=df.drop(columns=['TotalCall','customerID','PhoneService'])
```

```python
sample_df=['InternetService','OnlineSecurity','OnlineBackup','DeviceProtection','TechSupport']
for i in sample_df:
    print(df[i].unique())
```

```python
# These columns are only "No internet service" when InternetService == "No" — that's
# already captured by InternetService, so collapse it to "No" to avoid duplicate columns.
service_cols = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
                'TechSupport', 'StreamingTV', 'StreamingMovies']
for col in service_cols:
    df[col] = df[col].replace('No internet service', 'No')
```

```python
df.head()
```

```python
df_encoded = pd.get_dummies(df, drop_first=True)
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm
```

```python
X = df_encoded.drop(columns=['Churn']).astype(float)

# Add a constant column — statsmodels' VIF calculation expects this for correct results
X_with_const = sm.add_constant(X)

# Calculate VIF for each feature
vif_data = pd.DataFrame()
vif_data["Feature"] = X_with_const.columns
vif_data["VIF"] = [variance_inflation_factor(X_with_const.values, i) for i in range(X_with_const.shape[1])]

print(vif_data.sort_values("VIF", ascending=False))
```

```python
X = df_encoded.drop(columns=['Churn', 'VoiceMailPlan_Yes']).astype(float)

# Add a constant column — statsmodels' VIF calculation expects this for correct results
X_with_const = sm.add_constant(X)

# Calculate VIF for each feature
vif_data = pd.DataFrame()
vif_data["Feature"] = X_with_const.columns
vif_data["VIF"] = [variance_inflation_factor(X_with_const.values, i) for i in range(X_with_const.shape[1])]

print(vif_data.sort_values("VIF", ascending=False))
```

```python
X = df_encoded.drop(columns=['Churn', 'VoiceMailPlan_Yes','TotalRevenue']).astype(float)

# Add a constant column — statsmodels' VIF calculation expects this for correct results
X_with_const = sm.add_constant(X)

# Calculate VIF for each feature
vif_data = pd.DataFrame()
vif_data["Feature"] = X_with_const.columns
vif_data["VIF"] = [variance_inflation_factor(X_with_const.values, i) for i in range(X_with_const.shape[1])]

print(vif_data.sort_values("VIF", ascending=False))
```

```python
# Bin the zero-inflated NumbervMailMessages into None/Low/High instead of one
# continuous scale, so the model can learn a separate effect per group rather
# than assuming one smooth relationship across the whole range.
df['NumbervMailMessages_bin'] = pd.cut(
    df['NumbervMailMessages'],
    bins=[-1, 0, 20, 51],
    labels=['None', 'Low', 'High']
)
df = df.drop(columns=['NumbervMailMessages'])
```

```python
df['tenure_bin'] = pd.cut(
    df['tenure'],
    bins=[-1, 1, 23, 71, 72],
    labels=['1 month', '2-23 months', '24-71 months', '72 months (censored)']
)
df = df.drop(columns=['tenure'])
```
